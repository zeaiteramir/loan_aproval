```python
def detect_outlier(column):
    Q1 = np.quantile( column , 0.25)
    Q3 = np.quantile( column , 0.75)
    IQR = Q3 - Q1
    Lower = Q1 - 1.5*IQR
    Upper = Q3 + 1.5*IQR
    Outlier = column[ (  ( column < Lower) |  (column > Upper ) ) ]
    return Outlier
```

## Final pipeline (for Streamlit)
Preprocessing (ordinal-encode categoricals + MinMax scaling) and KNN in a single sklearn `Pipeline`, fitted on the training split only, then dumped with joblib.

```python
import joblib
from pathlib import Path
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OrdinalEncoder, MinMaxScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, recall_score, precision_score, f1_score

raw = pd.read_csv("data/loan_data.csv")
cat_cols = raw.select_dtypes("object").columns.tolist()
X_raw = raw.drop(columns="loan_status")
y_raw = raw["loan_status"]
feature_columns = X_raw.columns.tolist()

preprocess = ColumnTransformer(
    [("cat", OrdinalEncoder(handle_unknown="use_encoded_value", unknown_value=-1), cat_cols)],
    remainder="passthrough",
    verbose_feature_names_out=False,
).set_output(transform="pandas")

loan_pipeline = Pipeline([
    ("preprocess", preprocess),
    ("scale", MinMaxScaler()),
    ("knn", KNeighborsClassifier(n_neighbors=3)),
])

Xtr, Xte, ytr, yte = train_test_split(X_raw, y_raw, test_size=0.3, random_state=42)
loan_pipeline.fit(Xtr, ytr)
pred = loan_pipeline.predict(Xte)
metrics = {
    "accuracy": accuracy_score(yte, pred),
    "recall": recall_score(yte, pred),
    "precision": precision_score(yte, pred),
    "f1": f1_score(yte, pred),
}
print(metrics)

Path("models").mkdir(exist_ok=True)
joblib.dump(
    {
        "pipeline": loan_pipeline,
        "feature_columns": feature_columns,
        "categorical_options": {c: sorted(raw[c].unique().tolist()) for c in cat_cols},
        "numeric_ranges": {
            c: [float(raw[c].min()), float(raw[c].max()), float(raw[c].median())]
            for c in feature_columns if c not in cat_cols
        },
        "metrics": metrics,
    },
    "models/loan_pipeline.joblib",
)
print("saved models/loan_pipeline.joblib")
```
